**Exericio 2 D - Klein**

```python
import scipy as sp  
import numpy as np
import matplotlib as plt 
```

```python
TSR = 8.5 #razão velocidade de ponta 
P = 15 * 10**6 #Potencia da turbina em W
rho = 1.225 #massa especifica do ar em kg/m^3
V_0 = 10.59 # velocidade do ar nao perturbada em m/s 
L_D = 120  # razão lift/drag do perfil
```

```python
def cp_drag(x):

    coeffs = [16, -24, 9 - 3 * x**2, x**2 - 1]

    roots = np.roots(coeffs)

    a = [r.real for r in roots if abs(r.imag) < 1e-8 and 0.25 < r.real < 1/3][0]

    a_linha = (1 - 3 * a) / (4 * a - 1)

    return a_linha * (1 - a) * x**3 - (a_linha * (1 + a_linha) * x**4) / L_D


poly_int_drag = sp.integrate.quad(cp_drag, 1e-6, TSR)

Cp_drag = (8 / TSR**2) * poly_int_drag[0]
```

```python
R_drag = np.sqrt(P / (0.5 * rho * V_0**3 * Cp_drag * np.pi))
print(R_drag)
```
